```python
print('Exercício 079:')
continuar = ''
lista = []
igual = 0
while continuar != 'n':
    num = int(input('Digite o valor inteiro que deseja incerir a lista: '))

    #Loop para verificar se existem valores repetidos
    for i in range (len(lista[:])):
        if num == lista [i]:
            igual = 1
            break
        else:
            pass
            igual = 0

    #Verifica que se pode ou n colocar na lista
    if igual == 1:
        print('Valor presente na lista n vou adicionar!!')
    else:
        lista.append(num)
        print('Valar adicionado com sucesso!!!')
        igual = 0
    continuar = input('Deseja continuar?[S/N]: ').lower()

    #Verificando se a resposta esta coerente, [S/N]
    while continuar != 'n' and continuar != 's':
        if continuar != 'n' and continuar != 's':
            print('Opção selecionada n identificada, digite novamente!!')
        continuar = input('Deseja continuar?[S/N]: ').lower()

lista.sort() #Coloca em ordem crescente
print(f'A sua lista ficou assim: {lista}')
```

```python
print('Exercício 081:')  
l = []
continuar = 's'
cinco = 0
while continuar != 'n':
    num = int(input('Digite um valor para adicionar a lista: '))
    l.append(num)
    if num == 5:
         cinco += 1
    continuar = input('Deseja continuar?[S/N]: ').lower()
    while continuar != 'n' and continuar != 's':
            if continuar != 'n' and continuar != 's':
                print('Opção selecionada n identificada, digite novamente!!')
            continuar = input('Deseja continuar?[S/N]: ').lower()
l.sort(reverse=True)
print(f'A)Foram digitados {len(l[:])} numeros dentro da lista. \nB)A lista ordenada de forma decrescente: {l}. \nC)O valor 5 foi digitado: {cinco} vezes.')
```
