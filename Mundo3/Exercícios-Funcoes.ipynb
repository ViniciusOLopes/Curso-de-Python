```python
print('Exercício 097:')
print('═' * 40)

palavras = list()

def editando_lista(palavras):
    for palavra in palavras:
        tam = len(palavra)
        print('~' * (tam + 4))
        print(f'{'':>2}{palavra}')
        print('~' * (tam + 4))

for i in range(4):
    palavras.append(str(input(f'Digite a {i+1} palavra: ')))
    print('Palavra registrada com sucesso.')



editando_lista(palavras)
```

```python
from time import sleep
print('Exercício 098: ')
print('═' * 20)

def contagem(inicio, fim, passo):
    print('Iniciando a contagem: ')
    if inicio < fim:
        for i in range(inicio,fim, passo):
            print(f'{i}',end=', ')
            sleep(0.5)
        print(fim)
    else:
        for i in range(inicio, fim, -passo):
            print(f'{i}',end=', ')
            sleep(0.5)
        print(fim)
    print('')

contagem(1,10,1)
contagem(10,0,2)
contagem(0,50, 5)
```

```python
print('Exercício 099: ')
print('═' * 20)

def maior(*num):
    m = num[0]
    for i in range(len(num)):
        if m < num[i]: 
            m = num[i]
    print(f'Analisando números passados: {num}')
    print(f'O maior valor é {m}')


maior(4, 5,9, 1, 13, 43)
```
