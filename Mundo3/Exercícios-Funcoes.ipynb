```python
from time import sleep
print('Exercício 098: ')
print('═' * 15)

def contagem(inicio, fim, passo):
    print('Iniciando a contagem: ')
    if inicio < fim:
        for i in range(inicio,fim, passo):
            print(f'{i}',end=', ')
            sleep(0.5)
        print(fim)
    else:
        for i in range(inicio, fim, -passo):
            print(f'{i}',end=', ')
            sleep(0.5)
        print(fim)
    print('')

contagem(1,10,1)
contagem(10,0,2)
contagem(0,50, 5)
```

```python

print('Exercício 100: ')
print('═' * 15)

lista = list()
pares = list()

def sorteia(lista:list):
    from random import randint
    for i in range(5):
        lista.append(int(randint(0,100)))

def somar_par(lista:list,par:list):
    soma = 0
    for i in range(len(lista)):
        if lista[i]%2 == 0:
            soma += lista[i]
            par.append(lista[i])
    return soma


sorteia(lista)
soma = somar_par(lista, pares)

print(f'Os valores sorteados foram: {lista} \nA soma dos valores pares ({pares}): {soma}')
```

```python
print('Exercício 101: ')
print('═' * 15)

def voto(ano_nas):
    from datetime import datetime
    idade = datetime.now().year - ano_nas
    if idade < 18 and idade >= 0:
        return f'Com {idade} anos: NÃO VOTA'
    elif idade == 18 or (idade > 18 and idade < 65):
        return f'Com {idade} anos: VOTO OBRIGATORIO'
    elif idade >= 65:
        return f'Com {idade} anos: VOTO OPCIONAL'
    else:
        return f'Data de nascimento {ano_nas} inválida.'


nascimento = int(input('Digite a sua data de nascimento: '))
print(f'{voto(nascimento)}')

```

```python
print('Exercício 102: ')
print('═' * 15)

def fatorial(num: int, show=False):
    fator = 1
    retorno = ''
    for i in range(num):
        fator *= num -i
    if show:
        for i in range(num-1):
            retorno += f'{num} * '
            num-=1
        retorno += '1 = '
    return retorno + f'{fator}'
    

print(fatorial(8, False))
```

```python
print('Exercício 105: ')
print('═' * 15)

def calcular_notas(*notas, sit=False):
    num_notas = len(notas)
    menor = notas[0]
    maior = notas[0]
    media = 0
    situacao = list()
    dados = dict ()
    for i in range(num_notas):
        if menor > notas[i]:
            menor = notas[i]

        if maior < notas[i]:
            maior = notas[i]
        media += notas[i]

    media = media/num_notas
    informacoes = dict()
    informacoes['Quantidade de notas'] = num_notas
    informacoes['Maior nota'] = maior
    informacoes['Menor nota'] = menor
    informacoes['Media'] = f'{media:.1f}'
    
    if media >= 6:
        informacoes['situacao_turma'] = 'BOA'
    else:
        informacoes['situacao_turma'] = 'A BAIXO'
    
    if sit:
        for i in range(num_notas):
            dados['nota: '] = notas[i]
            if notas[i] >= 6:
                dados['situacao'] = 'APROVADO'
            else:
                dados['situacao'] = 'REPROVADO'
            
            situacao.append(dados.copy())
            dados.clear()
        informacoes['situacao'] = situacao
    return informacoes


resp = calcular_notas(5.5,8,10,3,6,7, sit= True)
print(resp)
```

```python
from time import sleep
print('Exercício 106: ')
print('═' * 15)

# Matriz de cores ANSI para o fundo do texto
# \033[ m -> Código de escape ANSI
CORES = {
    'sem_cor': '\033[m',
    'vermelho': '\033[0;30;41m',
    'verde': '\033[0;30;42m',
    'roxo': '\033[0;30;44m',
    'invertido': '\033[7;30m'
}


def titulo(msg, cor='sem_cor'):
    """Gera um título centralizado e estilizado com linhas adaptáveis."""
    tamanho = len(msg) + 4
    print(CORES[cor], end='')
    print('~' * tamanho)
    print(f'  {msg}  ')
    print('~' * tamanho)
    print(CORES['sem_cor'], end='')
    sleep(0.5)


def ajuda(comando):
    """Acessa o Interactive Help do Python e exibe o manual estilizado."""
    titulo(f"Acessando o manual do comando '{comando}'", 'roxo')
    print(CORES['roxo'], end='')
    help(comando)
    print(CORES['roxo'], end='')
    sleep(1)


# Programa Principal
while True:
    titulo('SISTEMA DE AJUDA INTERATIVA PYTHON', 'verde')
    
    comando = str(input('Função ou Biblioteca > ')).strip()
    
    # Condição de parada
    if comando.upper() == 'FIM':
        break
    else:
        ajuda(comando)

titulo('ATÉ LOGO!', 'vermelho')
```
