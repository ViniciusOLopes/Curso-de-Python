```python
from random import randint
from time import sleep

lista = list()
aux = dict()
jogadas = dict()

for i in range(4):
    jogadas['jogador'] = f'Jogador {i+1}'
    jogadas['dado'] = randint(1,6)
    lista.append(jogadas.copy())
    #print(jogadas)

for i in range(4):
    print(f'{lista[i]['jogador']} tirou: {lista[i]['dado']}')
    sleep(1)

for i in range(3,0,-1):
    j=0
    while j<i:
        if lista[j]['dado'] < lista[j+1]['dado']:
            aux = lista[j]
            lista[j]= lista[j+1]
            lista[j+1] = aux
        j+=1
print('Ordem de vitória: ', end='')
print(f'{lista[i]['jogador']} com: {lista[i]['dado']}')
for i in range(1,4):
    print(f'{' ':>18}{lista[i]['jogador']} com: {lista[i]['dado']}')
```
