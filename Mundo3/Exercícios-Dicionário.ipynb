```python
print('Exerício 091:')
print('═' * 15)

from random import randint
from time import sleep
print('Exerício 091:')
lista = list()
aux = dict()
jogadas = dict()

for i in range(4):
    jogadas['jogador'] = f'Jogador {i+1}'
    jogadas['dado'] = randint(1,6)
    lista.append(jogadas.copy())
    #print(jogadas)
print('═'*25)
for i in range(4):
    print(f'{lista[i]['jogador']} tirou: {lista[i]['dado']}')
    sleep(1)

for i in range(3,0,-1):
    j=0
    while j<i:
        if lista[j]['dado'] < lista[j+1]['dado']:
            aux = lista[j]
            lista[j]= lista[j+1]
            lista[j+1] = aux
        j+=1
print('═'*25)
print('Ordem de vitória: ')
print('═'*25)
for i in range(4):
    print(f'{i+1}° {lista[i]['jogador']} com {lista[i]['dado']}')
```

```python
print('Exercício 093:')
print('═' * 15)

dados = dict()
gols = list()

dados['nome'] = str(input('Nome do jogador: '))
jogos = int(input('Quantos jogos ele jogou no ultimo campionato? '))
dados['total_gols'] = 0

for i in range(jogos):
    gols.append(int(input(f'{'':>5}Quantos gols fora feitos no {i+1} jogo? ')))
    dados['total_gols'] += gols[i]
dados['gols'] = gols[:]

print('═' * 50)
print(dados)
print('═' * 50)

for k, v in dados.items():
    print(f'O campo {k} tem o valor: {v}')

print('═' * 50)
print(f'O jogador {dados["nome"]} jogou {jogos} partidas.')
for i in range(len(gols)):
    print(f'{'':>5} => Na partida {i+1}, fez {dados['gols'][i]} gols.')
print(f'Foi um total de {dados['total_gols']} gols.')
```

```python
print('Exercício 095:')
print('═' * 15)

jogadores = list()
dados = dict()
gols = list()

resp = 's'
cod = 0
jogador = 0
pego = 0


#Cadastramento de jogadores
while resp != 'n':
    dados['nome'] = str(input('Nome do jogador: '))
    dados['jogos'] = int(input('Quantos jogos ele jogou no ultimo campionato? '))
    dados['total_gols'] = 0

    #cadastro de gols para cada jogo
    for i in range(dados['jogos']):
        gols.append(int(input(f'{'':>5}Quantos gols fora feitos no {i+1} jogo? ')))
        dados['total_gols'] += gols[i]
    dados['gols'] = gols[:]
    dados['cod'] = cod
    cod += 1
    jogadores.append(dados.copy())

    #limpando os dados temporários
    dados.clear()
    gols.clear()

    
    print('Jogador cadastrado com sucesso!!!')
    print('═' * 50)
    resp  = input('Deseja continuar? [S/N]').lower()

print('Cod       nome         gols')
for j in range(len (jogadores)):
    print(f'{jogadores[j]['cod']}    {jogadores[j]['nome']},        gols: {jogadores[j]['total_gols']}') 

while jogador != 999:
    print('═' * 50)
    jogador = int(input('Mostrar dados de qual jogador?(999 encerrar) '))

    #Imprimir jogador selecionado:
    for i in range(len(jogadores)):
        if jogadores[i]['cod'] == jogador:
            print(f'══>LEVANTAMENTO DO JOGADOR {jogadores[i]['nome'].upper()}')
            for a in range(jogadores[i]['jogos']):
                print(f'No jogo {a+1} fez : {jogadores[i]['gols'][a]} gols.')
                pego = 1
    if pego == 0 and pego != 999:
        print('Codigo n identificado procure qual jogador deseja consultar: ')
        print('Cod       nome         gols')
        for j in range(len (jogadores)):
            print(f'{jogadores[j]['cod']}    {jogadores[j]['nome']},        gols: {jogadores[j]['total_gols']}')
    pego = 0

print('<<<<<OBRIGADO POR USAR O APP>>>>>') 
```
