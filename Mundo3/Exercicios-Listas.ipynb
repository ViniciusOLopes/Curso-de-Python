```python
print('Exercício 084:')
cadastro = list(list())
dados = list()
pesados = list()

cont = ''
while cont != 'n':
    dados.append(input('Digite um nome: '))
    dados.append(float(input('Digite seu peso: ')))        
    cadastro.append(dados[:])

    print(f'Pessoa Cadastrada com sucesso: {dados[:]}')
    dados.clear()

    cont = input('Deseja continuar? [S/N] ').lower()
print('=-' * 40)
print('Foram cadastrados {} pessoas;'.format(len(cadastro[:])))
print('=-' * 40)
#Usando o sorted para pegar apenas o segundo valor de cada pessoa da lista cadastro:
pesados = sorted(cadastro, key=lambda peso: peso[1], reverse = True)
for i in range(len(pesados)):
    print(f'A lista de mais Pesados: {pesados[i][0]}, {pesados[i][1]}Kg')
print('=-' * 40)
pesados.reverse()
for i in range(len(pesados)):
    print(f'A lista de mais leves: {pesados[i][0]}, {pesados[i][1]}Kg')
```

```python
from IPython.display import clear_output #Para dar clear atravez do código
from time import sleep
print('Exercício 089:')

#Variáveis
aluno = list()
dados = list()
notas = list()
média = 0
num_alunos = 0
achou = 0
matricula = 0

#Cadastro de alunos:
while True:
    print('═' * 50)
    print('Escolha uma das opções: \n[1]Cadastrar alunos ' \
                                  '\n[2]Consultar aluno' \
                                 ' \n[3]Imprimir bolim da turma ' \
                                 ' \n[4]Sair do programa' )
    resp = int(input('Digite sua escolha: '))
    match resp:

        #Cadastrar alunos
        case 1:
            clear_output()
            print('Cadastramento de alunos: ')
            num_alunos  = int(input('Digite o número de alunos que vc quer cadastrar: '))
            for i in range(num_alunos):
                nome = input('Nome do aluno:')
                notaA = float(input('Nota 1: '))
                notaB = float(input('Nota 2: '))

                dados.append(matricula)
                dados.append(nome)
                dados.append(notaA)
                dados.append(notaB)

                aluno.append(dados[:])
                dados.clear()
                matricula +=1

        #Escolher boletim por nome
        case 2:
            clear_output()
            print('═' * 50)
       
            consultar = int(input('Escreva a matrícula do aluno: '))
            print('Aluno sendo consultado...')
            sleep(2)
            for pessoa in range(len(aluno)):
                if aluno[pessoa][0] == consultar:
                    print('═' * 50)
                    print(f'{"BOLETIM ESCOLAR":^50}')
                    print('═' * 50)
                    print(f'{"Nº":<5} {"NOME":<18} {"N1":>5} {"N2":>5} {"MÉDIA":>8}')
                    print('─' * 50)
                    media = aluno[i][2] + aluno[i][3]
                    print(f'{aluno[i][0]:<5} {aluno[i][1]:<18} {aluno[i][2]:>5.1f} {aluno[i][3]:>5.1f} {media/2:>8.1f}')
                    achou = 1
            print('═' * 50)
            if achou != 1:
                print('Aluno n encontrado')
                achou = 0

        #Imprimir boletim completo
        case 3:
            clear_output()
            print('═' * 50)
            print(f'{"BOLETIM ESCOLAR":^50}')
            print('═' * 50)
            print(f'{"Nº":<5} {"NOME":<18} {"N1":>5} {"N2":>5} {"MÉDIA":>8}')
            print('─' * 50)
            for i in range(len(aluno)):
                media = aluno[i][2] + aluno[i][3]
                print(f'{aluno[i][0]:<7} {aluno[i][1]:<18} {aluno[i][2]:>5.1f} {aluno[i][3]:>5.1f} {media/2:>8.1f}')
            print('═' * 50)

        #Fechar programa
        case 4:       
              break

        #Digitando valor errado
        case _:
            print('Valor n identificado. escolha novamente.')
            pass
    if resp == 4:
        cont = input('Certeza que deseja sair do programa?[S/N]: ').lower()
        if cont == 'n':
                break
        elif cont == 's':
            pass
        else:
            print('Resposta n identificada. Cancelando execução do aplicativo.')
            break 
```
